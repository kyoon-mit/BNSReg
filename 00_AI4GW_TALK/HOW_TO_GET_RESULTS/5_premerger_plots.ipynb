```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams.update({
    'font.size': 12,
    'axes.spines.top': False,
    'axes.spines.right': False,
    'figure.dpi': 120,
})

df_frac = pd.read_csv(CSV_SNR_FRAC_SIG)
print(f'{len(df_frac):,} SNR bins  |  columns: {list(df_frac.columns)}')
df_frac.head(3)
```

## SNR vs Fraction Within Relative-Error Cutoff

```python
CUTS = [0.02, 0.05, 0.10, 0.50]
CLRS = ['#3d1a5e', '#7b2d8b', '#b85cae', '#e8a0d5']

snr_centers = df_frac['snr_bin'].values
ticks = list(range(5, 51, 5))

fig, ax = plt.subplots(figsize=(6, 4))

for cut, clr in zip(CUTS, CLRS):
    col = f'cutoff_{cut}'
    if col not in df_frac.columns:
        print(f'Column {col} not found, skipping')
        continue
    frac = df_frac[col].values * 100.0
    ax.plot(snr_centers, frac, lw=1.8, color=clr, label=f'{int(cut*100)}%')

ax.axhline(50, color='gray', ls='--', lw=1.0, alpha=0.5)
ax.set_xlabel('SNR')
ax.set_ylabel('% within cutoff')
ax.legend(frameon=False, fontsize=9, title_fontsize=10,
          title=r'$|(\hat{\mathcal{M}}_c - \mathcal{M}_c)/\mathcal{M}_c|$ <',
          loc='center right')
ax.set_xticks(ticks)
ax.set_xlim(snr_centers.min() - 0.5, snr_centers.max() + 0.5)
ax.set_ylim(0, 105)
fig.tight_layout()
plt.show()
```
