# Chirp Mass Histogram: Truth vs Background vs Signal

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ── paths ─────────────────────────────────────────────────────────────────────
BASE = Path("/n/holystore01/LABS/iaifi_lab/Lab/kyoon/BNSReg/outputs"
            "/s4d_gaussnll_15_30_pre_match_filter"
            "/two_masses_59s_63s-d64-s64-l4")

bg_path  = BASE / "background" / "background.csv"
sig_path = BASE / "signal"     / "signal.csv"

df_bg  = pd.read_csv(bg_path)
df_sig = pd.read_csv(sig_path)

print(f"Background rows : {len(df_bg)}")
print(f"Signal rows     : {len(df_sig)}")
```

```python
# ── bins (same as plot_mchirp_q_and_sigmas.py) ────────────────────────────────
BINS = np.linspace(0.5, 3.0, 51)   # 50 bins of width 0.05

# ── data ──────────────────────────────────────────────────────────────────────
mc_true = df_sig['M_chirp_true'].values   # truth from signal file
mc_bg   = df_bg['M_chirp_pred'].values    # background predictions
mc_sig  = df_sig['M_chirp_pred'].values   # signal predictions

# ── helper ────────────────────────────────────────────────────────────────────
def hist_outline(ax, data, bins, **kwargs):
    counts, edges = np.histogram(data, bins=bins)
    ax.stairs(counts, edges, **kwargs)

# ── plot ──────────────────────────────────────────────────────────────────────
plt.rcParams.update({'font.size': 11,
                     'axes.spines.top': False,
                     'axes.spines.right': False})

fig, ax = plt.subplots(figsize=(7, 4.5))

hist_outline(ax, mc_true, bins=BINS, color='black', linewidth=1.2, label='Truth')
hist_outline(ax, mc_bg,   bins=BINS, color='blue',  linewidth=1.2, label='Background')
hist_outline(ax, mc_sig,  bins=BINS, color='red',   linewidth=1.2, label='Signal')

ax.set_xlabel(r'$\mathcal{M}_{c}$')
ax.set_ylabel('Counts')
ax.legend(frameon=False)
fig.tight_layout()
plt.show()
```
